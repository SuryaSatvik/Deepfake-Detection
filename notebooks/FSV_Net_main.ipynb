# Superseded — open `FSV_Net_RUN_ALL.ipynb`

This file used to hold the multi-stage workflow. Everything it did now lives in
**`FSV_Net_RUN_ALL.ipynb`**, which is one notebook you run top to bottom:

* `DEMO_MODE = True`  — generates its own dataset, trains, evaluates, ablates and
  explains, with no download and no GPU. It ships with its outputs already in it, so
  you can read the results before running anything.
* `DEMO_MODE = False` — the same notebook against Celeb-DF / FaceForensics++ / DFDC.

Keeping two notebooks in step was a good way to end up with two different answers to
the same question, so there is now one.
